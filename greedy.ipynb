```python
import heapq

# return {
#     "objective": 102,
#     "schedule": [
#     task 0 -> processor 1, starts at 0, finishes at 3
#         (0, 1, 0, 3),
#         (1, 2, 0, 2),
#         (2, 1, 3, 7),
#         (3, 2, 3, 5),
#         (4, 2, 5, 6)
#     ]
# }
# while arr:
#     minimum = min(arr, key=lambda x: x[1])
#     arr.remove(minimum)

def greedy(instance):
    print(instance)
    objective = 0
    schedule = []
    tasks = instance['tasks']
    
    # return {
    #     "objective": 102,
    #     "schedule": [
    #     task 0 -> processor 1, starts at 0, finishes at 3
    #         (0, 1, 0, 3),
    #         (1, 2, 0, 2),
    #         (2, 1, 3, 7),
    #         (3, 2, 3, 5),
    #         (4, 2, 5, 6)
    #     ]
    # }
    # w * l

    processors = [(0, i) for i in range(instance['n_processors'])]
    heapq.heapify(processors)

    while tasks:
        current_time, processor_id = heapq.heappop(processors)
        available_tasks = [task for task in tasks if task[1] <= current_time]
        if (not available_tasks):
            # if there aren't any available tasks fetch the earliest released
            min_release = min(tasks, key=lambda x: x[1])[1]
            current_time = min_release
            earliest_released = [task for task in tasks if task[1] == min_release]
            task = max(earliest_released, key=lambda t: (current_time + t[2]) * t[3])
        else:
            task = max(available_tasks, key=lambda t: (current_time + t[2]) * t[3])

        tasks.remove(task)
        finish_time = current_time + task[2]
        schedule.append((processor_id, current_time, finish_time))
        heapq.heappush(processors, (finish_time + 1, processor_id))
        objective = objective + finish_time * task[3]
        
    return schedule, objective
        
    
```

```python
results = [greedy(instance) for instance in instances]
results
```
