```python
import csv
import numpy as np
import time

algos = [greedy, modified_greedy]
report_outs = ["reports/greedy.csv", "reports/modified_greedy.csv"]

header = ["dimension", "average_objective", "time"]

for i in range(2):
    with open(report_outs[i], "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(header)
        for k in range(4, 26):
            instances = dgu.read_instances(f"data/{k}_tasks.txt")
            start = time.perf_counter()

            results = [algos[i](instance) for instance in instances]
            objectives = [result[0] for result in results]
            average = np.average(objectives)

            end = time.perf_counter()
            duration = end - start
            writer.writerow([k, average, duration])
    
```
